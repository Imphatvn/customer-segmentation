Spending features are expected to be right-skewed. This is important because it affects both outlier detection and the choice between mean and median-based approaches.

## Age by Gender

A temporary age column is created for exploration. The final age feature is created in the preprocessing notebook.

```python
customer_info_eda = customer_info.copy()
customer_info_eda['customer_birthdate'] = pd.to_datetime(customer_info_eda['customer_birthdate'], errors='coerce')
customer_info_eda['age'] = 2025 - customer_info_eda['customer_birthdate'].dt.year

plot_age_by_gender(customer_info_eda)
```

## Complaints and Loyalty Over Time

The complaint and loyalty-card variables are temporarily converted into binary indicators to calculate rates by first-transaction year.

```python
customer_info_eda['complained'] = (customer_info_eda['number_complaints'].fillna(0) > 0).astype(int)
customer_info_eda['loyalty_card'] = customer_info_eda['loyalty_card_number'].notna().astype(int)

plot_complaints_and_loyalty_over_time(customer_info_eda)
```

## Promotions and Education

The promotion percentage contains invalid values, so it is temporarily corrected here for visualization. The permanent correction is applied during preprocessing.

```python
customer_info_eda['degree'] = customer_info_eda['customer_name'].str.contains('Phd.|Msc.|Mst.|Bsc.', case=False, na=False).astype(int)

promotion = customer_info_eda['percentage_of_products_bought_promotion'].abs()
promotion[promotion > 1] = promotion[promotion > 1] / 10
customer_info_eda['pct_prod_in_promotion'] = promotion

plot_promotion_by_degree(customer_info_eda)
```

## Spending Relationships

The spending variables are analysed using their lifetime values. Strong correlations may indicate redundant information that can later be reduced with PCA or autoencoders.

```python
spend_columns = [
    'lifetime_spend_groceries',
    'lifetime_spend_electronics',
    'lifetime_spend_vegetables',
    'lifetime_spend_nonalcohol_drinks',
    'lifetime_spend_alcohol_drinks',
    'lifetime_spend_meat',
    'lifetime_spend_fish',
    'lifetime_spend_hygiene',
    'lifetime_spend_petfood',
    'lifetime_spend_videogames',
]

plot_spend_correlation(customer_info_eda, spend_columns)
```

## Spending by Age Group

Annual spending cannot be calculated before preprocessing because `customer_lifetime` has not yet been created. For EDA, lifetime grocery spending is used as an initial indicator of spending behaviour.

```python
customer_info_eda['age_group'] = pd.cut(
    customer_info_eda['age'],
    bins=[22, 35, 50, 65, float('inf')],
    labels=['23-35', '36-50', '51-65', '66+'],
)

age_spend = (
    customer_info_eda
    .groupby('age_group', observed=True)['lifetime_spend_groceries']
    .mean()
    .reset_index()
)

import plotly.express as px

fig = px.bar(
    age_spend,
    x='age_group',
    y='lifetime_spend_groceries',
    text_auto=',.0f',
    color='lifetime_spend_groceries',
    color_continuous_scale='Blues',
    width=900,
    height=500,
)

fig.update_traces(textposition='outside')
fig.update_layout(
    title='Average Lifetime Grocery Spending by Age Group',
    xaxis_title='Age Group',
    yaxis_title='Average Lifetime Grocery Spending',
    coloraxis_showscale=False,
)
fig.show()
```

## Main Observations

- The dataset contains 34,060 customers and 25 features.
- Missing values are concentrated in loyalty-card number, birth date, complaints, children at home, and some spending categories.
- The promotion percentage has values below zero and above one, confirming a data-quality issue.
- Spending variables are right-skewed, which means extreme values should be handled carefully before clustering.
- Several spending categories are positively correlated, suggesting potential for dimensionality reduction.
- Complaint rates remain high over time, while loyalty-card ownership appears lower among more recent customers.
